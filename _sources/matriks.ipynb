# Matrik
## Definisi Matrik
Matriks adalah susunan bilangan, simbol, atau ekspresi yang diatur dalam baris dan kolom sehingga membentuk jajaran persegi panjang. Bilangan-bilangan yang tersusun dalam matriks disebut elemen atau anggota matriks. 

## Operasi Aritmatika Matrik
Terdapat tiga macam bentuk operasi aljabar pada matriks, yaitu operasi penjumlahan, pengurangan, dan perkalian.


## Penjumlahan Matrik
### Konsep penjumlahan matrik
Konsep penjumlahan matriks adalah operasi matematika yang melibatkan penjumlahan elemen-elemen matriks yang seletak dari dua matriks atau lebih yang memiliki ordo yang sama. Hasil penjumlahannya adalah matriks baru yang memiliki ordo yang sama dengan matriks semula. 

## Perkalian Matrik
### Konsep perkalian matrik
Dalam matematika, perkalian matriks adalah suatu operasi biner dari dua matriks yang menghasilkan sebuah matriks. Agar dua matriks dapat dikalikan, banyaknya kolom pada matriks pertama harus sama dengan banyaknya baris pada matriks kedua.

## Pengurangan Matrik
### Konsep pengurangan matrik
Konsep pengurangan matriks adalah mengurangkan elemen-elemen matriks yang seletak dari dua matriks yang memiliki ordo yang sama. 

```python
import numpy as np

# Definisi matriks A dan B
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

# Cetak matriks awal
print("Matriks A:")
print(A)
print("\nMatriks B:")
print(B)

# Penjumlahan matriks
sum_matrix = A + B
print("\nPenjumlahan Matriks:")
print(sum_matrix)

# Pengurangan matriks
diff_matrix = A - B
print("\nPengurangan Matriks:")
print(diff_matrix)

# Perkalian matriks (dot product)
mul_matrix = np.dot(A, B)
print("\nPerkalian Matriks:")
print(mul_matrix)
```

1. Determinan Matriks
Definisi:
Determinan matriks adalah nilai skalar yang dapat dihitung dari elemen-elemen matriks persegi (matriks yang jumlah baris dan kolomnya sama). Determinan matriks sering dilambangkan dengan det(A) atau ∣A∣

### Matriks 2x2:

Untuk matriks $A$ berukuran $2\times2$

$$A=\begin{pmatrix}a&b\\c&d\end{pmatrix}$$

Determinan matriks $A$ dihitung dengan rumus:

$$\det(A)=ad-bc$$


Contoh:
Misalkan matriks $A=\begin{pmatrix}3&2\\1&4\end{pmatrix}$,maka determinannya adalah


$$\det(A)=(3\times4)-(2\times1)=12-2=10$$

\begin{gathered}\mathrm{Matriks~3\times3:}\\UntukmatriksAberukuran3\times3\\A={\begin{pmatrix}a&b&c\\d&e&f\\g&h&i\end{pmatrix}}\\DeterminanmatriksAdihitungdenganrumus:\\\operatorname*{det}(A)=a(ei-fh)-b(di-fg)+c(dh-eg)\\Contoh:\\\text{Misalkan matriks }A=\begin{pmatrix}1&2&3\\4&5&6\\7&8&9\end{pmatrix},\mathrm{maka~determinannya~adalah;}\\\mathrm{t}(A)=1(5\times9-6\times8)-2(4\times9-6\times7)+3(4\times8-5\times7)\\=1(45-48)-2(36-42)+3(32-35\\=1(-3)-2(-6)+3(-3)\\=-3+12-9=0\end{gathered}

Sifat-Sifat Determinan:


1.Jika matriks $A$ memiliki satu baris atau kolom yang semua elemennya nol, maka $\det(A)=0.$
2. Jika dua baris atau kolom matriks $A$ identik, maka $\det(A)=0.$
3. Jika matriks $A$ adalah matriks segitiga (atas atau bawah), maka determinannya adalah hasil kali elemen-
elemen diagonal utama.
4.Jika matriks $A$ dan $B$ berukuran sama, maka $\det(AB)=\det(A)\times\det(B).$

## 2.Invers Matriks

### Definisi:

Invers matriks $A$ adalah matriks $A^{-1}$ sedemikian sehingga:
$$A\times A^{-1}=A^{-1}\times A=I$$

di mana $I$ adalah matriks identitas

### Syarat Invers Matriks:

Matriks $A$ memiliki invers jika dan hanya jika $\det(A)\neq0.$

Sifat-Sifat Invers Matriks:


1. Jika $A$ memiliki invers, maka inversnya unik.
2. Jika $A$ dan $B$ memiliki invers, maka $(AB)^{-1}=B^{-1}A^{-1}.$
3.Jika $A$ memiliki invers, maka $(A^{-1})^{-1}=A.$
4, Jika $A$ memiliki invers, maka $(A^T)^-1=(A^{-1})^T.$
